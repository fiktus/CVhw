# Обработайте данные для того, чтобы к ним можно было применить LogisticRegression. Т.е. отнормируйте числовые признаки, а категориальные закодируйте с помощью one-hot-encoding'а.

```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import make_pipeline
```

```python

X_train, X_test, y_train, y_test = train_test_split(data[feature_cols], data[target_col], random_state = 41, shuffle = True)
```

```python
a = [float(i) for i in X_train['TotalSpent'] if i != ' '] 
mn = round(np.mean(a), 3)
X_train['TotalSpent'] = X_train['TotalSpent'].replace(' ', mn)
X_test['TotalSpent'] = X_test['TotalSpent'].replace(' ', mn)
```

```python
a = StandardScaler()
X_train[num_cols] = a.fit_transform(X_train[num_cols])
X_test[num_cols] = a.transform(X_test[num_cols])

onehot = OneHotEncoder(sparse_output = False)
onehot.fit(X_train[cat_cols])
names = onehot.get_feature_names_out(cat_cols)
X_tr = pd.DataFrame(onehot.transform(X_train[cat_cols]), columns=names, index=X_train.index)
X_ts = pd.DataFrame(onehot.transform(X_test[cat_cols]), columns=names, index=X_test.index)
```

```python
X_train = X_train.drop(cat_cols, axis = 1)
X_test = X_test.drop(cat_cols, axis = 1)
X_train = pd.concat([X_train, X_tr], axis = 1)
X_test = pd.concat([X_test, X_ts], axis = 1)
```

# С помощью кроссвалидации или разделения на train/valid выборку протестируйте разные значения гиперпараметра C и выберите лучший (можно тестировать С=100, 10, 1, 0.1, 0.01, 0.001) по метрике ROC-AUC.

```python
from sklearn.metrics import roc_auc_score

cs = [100, 10, 1, 0.1, 0.01, 0.001]
roca = {}
for i in cs:
    logregl = LogisticRegression(C = i, random_state = 41, max_iter = 1000)
    ft = logregl.fit(X_train, y_train)
    
    y_predl = logregl.predict_proba(X_test)[:, 1]
    roc_aucl = roc_auc_score(y_test, y_predl)
    roca[i] = roc_aucl
```

# Протестируйте разные занчения параметров количества деревьев и learning_rate'а и выберите лучшую по метрике ROC-AUC комбинацию.

```python
score = {}
lr = [0.3, 0.1, 0.05, 0.03, 0.01]
trees_c = [100, 500, 1000, 2000]
best = 0.
for l in tqdm(lr):
    score[l] = {}
    for t in trees_c:
        boost = CatBoostClassifier(cat_features = cat_cols, learning_rate = l, iterations = t, auto_class_weights='Balanced' )
        boost.fit(X_trainb, y_trainb, verbose=False)
        y_predb = boost.predict_proba(X_testb)[:, 1]
        roc_aucb = roc_auc_score(y_testb, y_predb)

        best = max(best, roc_aucb)
        score[l][t] = roc_aucb
```

```python
print(best)
score
```

```python
boost = CatBoostClassifier(cat_features = cat_cols, learning_rate = 0.101, iterations = 100, auto_class_weights='Balanced', l2_leaf_reg = 1.5, border_count = 300)
boost.fit(X_trainb, y_trainb, verbose=False)
y_predb = boost.predict_proba(X_testb)[:, 1]
roc_aucb = roc_auc_score(y_testb, y_predb)
```
